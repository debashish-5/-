To clear things up: Metadata is never deleted or discarded during the embedding process. It lives alongside your vectors to give them context.
Here is the breakdown of what happens at each stage:
## Q1: Is metadata discarded/deleted?
No. When you "embed" a document, you are only converting the page_content (the text) into a numerical vector so the computer can understand the meaning [1, 2]. The metadata (like medicine or source) stays in its original text/JSON format. It is stored in a separate data structure (often a document store or a dictionary) that is linked to that specific vector [1]. [1] 
## Q2: When storing in FAISS, what is stored?
FAISS itself is a vector-only database. It technically only stores the high-dimensional numbers (the embeddings) and an ID [2].

* The Embedding: This is what FAISS uses to "search."
* The Link: Most frameworks (like LangChain or LlamaIndex) manage a "store" where the ID in FAISS points to your original Document object (which contains your important metadata) [3].

## Q3: When retrieved, what happens?
The retrieval process works in two steps:

   1. Vector Search: You input a query, it gets converted to a vector, and FAISS finds the most similar vectors based on math [2].
   2. Reconstruction: Once FAISS finds the best "ID," the system looks up that ID in the document store and hands you back the entire Document object—including your page_content and all your metadata [1, 3].

Why this matters for you:
Since you mentioned metadata is important, you can use it for Metadata Filtering. For example, you can tell your retriever: "Find the most relevant document, but only search through documents where medicine is 'Sticky Trap'."

# is metadata when embedding do ,...is embedding happened of metadata or is it store in faiss

When you perform embedding, the embedding is only performed on the page content, not the metadata. However, the metadata is not discarded; it is stored separately to provide context during retrieval. [1, 2, 3, 4] 
## 1. What happens during Embedding?

* Vectorization: The embedding model converts only the page_content text into a numerical vector (the embedding).
* Metadata Status: Metadata remains as raw text or JSON. It is not converted into a vector unless you manually combine it with your content before embedding. [2, 5, 6] 

## 2. How is it stored?
FAISS itself is a high-performance library that strictly manages vector indices. It does not natively store text or metadata. [2, 7, 8, 9] 

* FAISS Index: Stores the numerical embeddings and assigns each one a unique ID.
* Document Store (Docstore): When using a framework like [LangChain](https://docs.langchain.com/oss/python/integrations/vectorstores/faiss), a separate "docstore" (like an InMemoryDocstore) is created. This store maps the FAISS ID back to your original Document object, which contains both the page_content and the metadata. [2, 4, 10, 11, 12] 

## 3. What happens during Retrieval?

   1. Search: You provide a query, which is embedded into a vector. FAISS finds the most similar vector IDs.
   2. Lookup: The system uses those IDs to look up the corresponding documents in the Docstore.
   3. Return: You receive the full Document object, including all your important metadata. [1, 4, 10, 13] 

## Summary Table

| Component [2, 5, 11] | What is it? | Stored in FAISS Index? | Stored in Docstore? |
|---|---|---|---|
| Embedding | Numerical vector of page_content | Yes | No |
| Page Content | The original text | No | Yes |
| Metadata | Extra info (source, medicine, etc.) | No | Yes |

To ensure your metadata is searchable, you can use Metadata Filtering (e.g., filter={'medicine': 'Sticky Trap'}) in your LangChain Similarity Search. [5, 14]
